## Disease Background: 

* Prevalence & incidence: Worldwide, Alzheimer's cases have been estimated at 33-38 million cases with Alzheimer's making up roughly 60-70% of the 55 million worldwide cases of dementia. According to this source(below), dementia cases are estimated to reach 78 million people in 2030 and 139 million in 2050, with estimates of Alzheimer's cases to reach as much as 54.6 and 97.3 million. In America, 7.4 million people are estimated to live with Alzheimer’s with cases by 2060 to reach an estimated 13.8 million people (source:https://www.alz.org/alzheimers-dementia/facts-figures#prevalence).
* Economic burden: There is a heavy economic burden associated with Alzheimer’s Disease because of long term medical care required. According to this source(below), the costs for people living with dementia are estimated at $409 billion. In 2025, unpaid caregivers provided more than 19 billion hours of care valued at more than $446 billion. Source:https://www.alz.org/alzheimers-dementia/facts-figures?utm_source=chatgpt.com
* Risk factors (genetic, lifestyle): The biggest known risk factor for Alzheimer’s disease is age. Most cases of Alzheimer’s occur after age 65 with one in 13 people aged 65-84 and one in three people aged 85 and over having Alzheimer’s. Genetics also plays a factor in the development of Alzheimer’s as people with parents or siblings diagnosed with Alzheimer's have a higher risk of developing the disease. Though not directly linked, it is believed that high blood pressure, cardiac diseases, sleep deprivation, unhealthy diet, and lack of mental stimulation can increase the risk of Alzheimer's (source:https://www.nia.nih.gov/health/alzheimers-causes-and-risk-factors/thinking-about-your-risk-alzheimers-disease-five?utm_source=chatgpt.com).
* Societal determinants:There are multiple social determinants including social engagement (regularly seeing friends and family), sleep (getting adequate sleep), personality traits (being more outgoing), higher levels of education, and increased  length of employment. Source: https://www.nia.nih.gov/10-years-alzheimers-disease-and-related-dementias-research/increased-understanding-how-social?utm_source=chatgpt.comsed
* Symptoms:Alzheimer’s is most commonly associated with memory loss but has many other cognitively deteriorating effects such as thinking, reasoning, and behavioral ability which interfere with daily life. As Alzheimer’s progresses, people eventually lose their ability to perform routine tasks such as eating and walking. Early symptoms of Alzheimer’s are usually memory problems such as finding the right words to say, reasoning problems, or general forgetfulness. These cognitive problems continue until basic tasks become difficult for Alzheimer’s patients causing them to need help with everyday functions (source:https://www.nia.nih.gov/health/alzheimers-and-dementia/what-alzheimers-disease?utm_source=chatgpt.com).
* Diagnosis:The first course of action is a blood-based biomarker test. These tests look for presence of beta-amyloid plaques in the blood and APOE status (see genetics). There is also imaging that can be done as the disease progresses such as PET and MRI scans. (source:https://www.nia.nih.gov/10-years-alzheimers-disease-and-related-dementias-research/discovering-tools-detect-diagnose?utm_source=chatgpt.com)
* Standard of care treatments (& reimbursement): Although there are no current medications to cure Alzheimer’s, there are several FDA approved drugs to help manage the disease. Cholinesterase is a drug which helps to combat the breakdown of acetylcholine, a neurotransmitter which is believed to be important for memory and thinking, to combat cognitive impairment. As Alzheimer’s progresses though cholinesterase becomes less effective. Lecanemab and donanemab are two other drugs which combat beta-amalyloid, a protein which creates plaques which are believed to contribute to Alzheimer’s. Insurance only covers these drugs in certain situations with Medicare Part B covering part of the cost of these medications for patients who meet the criteria. Memantine is used to treat moderate to severe Alzheimer’s as it can help patients to maintain basic functions, such as going to the bathroom, as the disease progresses (source:https://www.nia.nih.gov/health/alzheimers-treatment/how-alzheimers-disease-treated?utm_source=chatgpt.com). Alzheimer’s disease often requires extensive caregiver support and is only covered by insurance in specific situations.
* Disease progression & prognosis:People with Alzheimer’s Disease (AD) experience memory loss as well as cognitive impairment. As the disease progresses, memory loss and cognitive impairment worsens until death. It causes death by reducing cognition until patients can no longer conduct basic vital functions such as swallowing which causes aspiration pneumonia.(source:https://www.nia.nih.gov/about/budget/alzheimers-disease-and-neuroscience-aging-1?utm_source=chatgpt.com)
* Continuum of care providers:Due to the debilitating nature of Alzheimer’s, care often needs to spread beyond hospitals and medicine. When first diagnosed with Alzheimer’s, it is recommended to consult a doctor about next steps and possible medical treatments. Since Alzheimer’s can impair basic function as the disease progresses, it is also recommended to explore care options such as assisted living. It is also recommended to set up family care, social workers, and community programs. Finally, it is recommended for people with Alzheimer’s to continue to live a healthy lifestyle including eating healthy and exercise (source:https://www.nia.nih.gov/health/alzheimers-treatment/next-steps-after-alzheimers-diagnosis?utm_source=chatgpt.com).
* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology):The disease starts in the synapses of the brain. It begins with the formation of amyloid beta plaques which block the synaptic gap between neurons inhibiting communication between them. Among these are two types of plagues, amyloid beta 40 and 42. An increased amount of 42 indicates Alzheimer’s disease. Additionally, the second part of this disease is the formation of tau tangles. These form within the neuron cell and block the axon affecting the ability for an action potential to travel down. These two protiens have debilitating effects on the function of neurons and their ability to communicate with other neurons. This diesase targets the regions of the brain associated with memory and cognition, causing memory loss and coginitive impairment. Source is Dr. Pierce-Cottler’s lecture.
* Clinical Trials/next-gen therapies:Looking into the future, researchers are working on developing preventative measures and cures for Alzheimer’s disease. Current preventative studies are trying to find links between genetic markers and Alzheimer’s disease diagnoses in order to administer preventative drugs to combat accumulation of amyloid beta and tau, two biomarkers which are believed to play a role in Alzheimer’s. The hope is to find people at high risk for Alzheimer’s to give these drugs to in order to hopefully delay the onset of the disease (source:https://clinicaltrials.gov/study/NCT05269394?utm_source=chatgpt.com). Currently there is no cure for Alzheimer’s, but research targeted and genetic risk factors such as the APOE2 gene, amyloid beta, and tau are creating hope that the effects and development of Alzheimer’s disease can be slowed and eventually cured.
